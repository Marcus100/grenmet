# Unidata Python notebook collection

**1010 example notebooks from 30 official repositories.** Each original notebook or Python file stays separate; the tables below group them by repository, episode, and feature. Python scripts have a sibling `filename.py.ipynb` notebook; original files remain byte-for-byte intact.

MetPy Mondays: **304 episodes with collected code**, **34 with no source file found**, across **338 inventory entries**. Entries such as `183b` are preserved. The sheet's status is shown as recorded; local availability is checked against actual files.

## Start here

1. Browse **MetPy Mondays by episode** below, or use the repository table for a topic or workshop.
2. Open one example notebook. Its original directory structure and small supporting assets are preserved in `sources/`.
3. Read that repository's README and environment/requirements file. Archived examples may need older package versions and unavailable remote/local datasets. This is a collected reference library, not an assertion that all examples run with the current workspace environment.

Launch from `/workspace/notebooks` using `uv run --frozen --package notebooks jupyter lab`, then open `metpy/unidata.ipynb`. The existing `cartopy.ipynb` remains available separately.

No upstream example code has been run. Some examples send messages, upload files, start servers or require accounts; inspect the relevant example before running it. Index searches below only read local metadata.

## Search by topic

Edit `query` below to find notebook titles and paths, for example `skew`, `radar`, `satellite`, `hurricane`, `qg`, or `xarray`.

```python
import json
from pathlib import Path
from urllib.parse import quote
from IPython.display import Markdown, display

query = "skew"
collection_root = next(
    p for p in (Path.cwd(), *Path.cwd().parents, Path.cwd() / "metpy")
    if (p / "manifest.json").is_file() and (p / "unidata.ipynb").is_file()
)
collection = json.loads((collection_root / "manifest.json").read_text())
matches = [e for e in collection["examples"]
           if query.casefold() in (e["title"] + " " + e["notebook"]).casefold()]
display(Markdown("\n".join(
    f"- [{e['title'].replace('[', '').replace(']', '')}]({quote(e['notebook'])}) — {e['repository']}"
    for e in matches
) or "No matches."))
```
